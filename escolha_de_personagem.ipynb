```python
#dados
class personagens:
  def __init__(self,clas,nome,arrem,posi,dib,defe,vel,passe):
   self.clas = clas
   self.nome = nome
   self.arrem = arrem
   self.posi = posi
   self.dib = dib
   self.defe = defe
   self.vel = vel
   self.passe = passe
  def status(self):
    return(f"[{self.clas}] nome:{self.nome} arremeso: {self.arrem} posicionamento: {self.posi} dible: {self.dib} defesa: {self.defe} velocidade: {self.vel} passe: {self.passe} \n")
  def apre(self):
    return(f"\n As caracteristcas do seu personagem e: {self.status()} \n ")

natan = personagens(1,"Natan",2,3,2,5,4,5)
kauan = personagens(2,"Kauan",1,4,4,4,5,3)
renato = personagens(3,"Renato",5,4,3,3,3,3)
joão = personagens (4,"João",4,5,2,2,3,5)

#codigo
print("""BASQUETE DA JUVETUDE

AVISO: PARA MELHOR EXPERIÊNCIA, PEÇO QUE SIGA O PROGRAMA CORRETAMENTE. ,OBRIGADO. \n

HISTORIA:
Um grupo de amigos com 4 integrantes decidiu fazer uma partida de basquete entre si,
2 vs 2 valendo pagar um rodizio,escolhar seu time e tenter ganha essa partida,boa sorte
""")

print("Os personagens são: ")
print(natan.status(),
      kauan.status(),
      renato.status(),
      joão.status()
      )

print("Escolhar seu personagem de 1 a 4: [1]natan(defesa) [2]kauan(velocidade) [3]joão(posicionamento) [4]renato(arremeso): ")

escolhar = int (input())
if escolhar == 1:
  print(natan.apre())
  prota = natan
elif escolhar == 2:
    print(kauan.apre())
    prota = kauan
elif escolhar == 3:
    print(joão.apre())
    prota = joão
elif escolhar == 4:
    print(renato.apre())
    prota = renato
else:
 print("personagem invalido")

print("Agora escolha a sua dupla: [1]natan [2]kauan [3]joão [4]renato")
dupla = int (input())

if dupla == escolhar:
  print("Esse ja e seu personagem")
elif dupla == 4:
  print(f"{renato.nome} e so passa a bola que e cesta\n")
  aliado = renato
elif dupla == 2:
  print(f"\n{kauan.nome} vai correr o campo todo\n")
  aliado = kauan
elif dupla == 3:
  print(f"{joão.nome} vai esta sempre no melhor lugar em campo\n")
  aliado = joão
elif dupla == 1:
  print(f"{natan.nome} vai impedir seus oponetes de marcar\n")
  aliado = natan
else:
  print("invalido")

print(f"Time formando {prota.nome} e {aliado.nome} hora da ação")

if (escolhar == 1  and dupla ==  2) or (escolhar == 2 and dupla == 1):
  print(f"Seus oponente são: {renato.nome} e {joão.nome}\n")
  inimigo1 = renato
  inimigo2= joão
elif (escolhar == 1  and dupla == 3) or (escolhar ==  3 and dupla == 1):
  print(f"Seus oponente são: {kauan.nome} e {joão.nome}\n")
  inimigo1 = kauan
  inimigo2= joão
elif (escolhar == 1 and dupla == 4) or (escolhar ==  4 and dupla == 1):
  print(f"Seus oponente são: {kauan.nome} e {renato.nome}\n")
  inimigo1 = renato
  inimigo2= kauan
elif (escolhar ==  2 and dupla == 3) or (escolhar == 3 and dupla == 2):
  print(f"Seus oponente são: {natan.nome} e {joão.nome}\n")
  inimigo1 = natan
  inimigo2= joão
elif (escolhar == 3 and dupla == 2) or (escolhar == 2 and dupla ==  3):
  print(f"Seus oponente são: {natan.nome} e {renato.nome}\n")
  inimigo1 = renato
  inimigo2= natan
elif (escolhar == 4 and dupla == 3) or (escolhar == 3 and dupla == 4):
  print(f"Seus oponente são: {natan.nome} e {kauan.nome}\n")
  inimigo1 = natan
  inimigo2= kauan
else:
  print("invalido")
#fim da primeira parte inicio da segunda

#dados
troca = (prota.passe * aliado.passe) - (inimigo1.posi * inimigo2.posi)
cesta = (prota.arrem * aliado.arrem) - (inimigo1.defe * inimigo2.defe)
bloqueio = (prota.defe * aliado.defe) - (inimigo1.passe * inimigo2.passe)
area =  (prota.posi * aliado.posi) - (inimigo1.posi * inimigo2.posi)
esquiva = (prota.dib * aliado.dib) - (inimigo1.vel * inimigo2.vel)
correr =  (prota.vel * aliado.vel) - (inimigo1.vel * inimigo2.vel)

time1 =  0
time2 =  0

#codigo

print("""TIMES FORMADOS E POSICIONADOS PARA QUE O JOGO COMECE.

COMO FUNCIONA:
voce tera algumas situações com duas opções que depededo das caracteristcas do seu time
contra seus inimigos pode acarreta e situações favoraveis ou negativas para voces
jogos acabar apos todas as situações acontecerem e o time que fizer mais cesta ganhar

HORA DO JOGO: \n
""")

print("Você esta com a bola [1]passar ou [2]arremessar \n")
decisão =  int (input())
if (decisão == 1 and troca >=4 and troca<=5):
 print("Fez passe com sucesso")
elif decisão ==1 and troca >=6:
 print(f"O {aliado.nome} fez cesta")
 time1 =+1
elif decisão == 1 and troca <= 3 and troca >=-2:
 print("Passe interceptado")
elif decisão ==1 and troca <=-3:
 print("Seus adversarios impediram seu passe e fizeram uma cesta")
 time2 =+1

elif decisão ==2 and cesta >=6:
 print("Fez a cesta")
 time1 =+1
elif decisão ==2 and cesta <= 5 and cesta >=-2:
 print("Você errou a cesta")
elif decisão ==2 and cesta <=-3:
 print("Você erro e seus advesarios aproveitaram a sobra e fizeram cesta")
 time2 =+1
else:
  print ("invalido")

print("Os oponente estão com a bola [1]defeder ou [2]posiciona")
decisão =  int (input())
if decisão == 1 and bloqueio >=4 and bloqueio <=5:
  print("Defedeu")
elif decisão == 1 and bloqueio >=6:
  print(f"Você defedeu e {aliado.nome} aproveito a breça e fez a cesta")
  time1 =+1
elif decisão == 1 and bloqueio <= 3 and bloqueio >=-2:
  print("Não conseguiu bloquear")
elif  decisão == 1 and bloqueio <=-3:
  print("eles abriram espaço e seus advesarios fizeram uma cesta")
  time2 =+1
elif decisão == 2 and area >=4 and area <=5:
  print("Você bloqueio o caminho")
elif decisão == 2 and area >=6:
  print(f"Você fez ele erra {aliado.nome} aproveito a breça e fez cesta")
elif decisão == 2 and area <= 3 and area >=-2:
  print("Eles passaram por voce")
elif decisão == 2 and area <=-3:
  print("Você não chega a tempo e seus advesarios fizeram cesta")
  time2= +1
else:
  print("invalido")

print("Você esta com a bola no meio de campo e seu oponente vem ter marca,voce [1]dibra ou tentar vencer na [2]velocidade")
decisão =  int (input())
if decisão ==1 and esquiva >=4 and esquiva<=5:
  ("Você dibra com sucesso o oponete")
elif decisão == 1 and esquiva >=6:
  print("Você avança e faz a cesta")
  time1  =+1
elif decisão == 1 and esquiva <= 3 and esquiva >=-2:
  print("Você tenta mas eles rouba a bola de você")
elif decisão == 1 and esquiva <=-3:
  print("E uma troca de passes rapida eles fazer cesta")
  time2 =+1

elif decisão == 2 and correr >=4 and correr<=5:
  print("Você correr mais rapido que ele")
elif decisão == 2 and correr >=6:
  print("Corredo como o vento você faz a cesta")
  time1 =+1
elif decisão == 2 and correr <= 3 and correr >=-2:
  print("Você esta muito lento")
elif decisão == 2 and correr <=-3:
  print("Eles rouba a bola de você e fazer a cesta")
  time2 =+1
else:
  print("invalido")

print("Sua dupla esta com a bola,você vai ser [1]posiciona para receber ou [2]defeder caso a bola sobre")
decisão =  int (input())
if decisão == 1 and area >=4 and area <=5:
  ("Você se posiciona perfeitamente para receber o passe")
elif decisão == 1 and area >=6:
  print("Você fez cesta")
  time1 =+1
elif decisão == 1 and area <= 3 and area >=-2:
  print("Você não conseguer achar uma boa posição")
elif decisão == 1 and area <=-3:
  print("seus advesarios intercepta a bola e fazer a cesta")
  time2 =+1

elif decisão == 2 and bloqueio >=4 and bloqueio<=5:
  print("A bola sobrou e você conseguiu defeder")
elif decisão == 2 and bloqueio >=6:
  print("Mesmo sem argulho você faz a cesta")
  time1 = +1
elif decisão == 2 and bloqueio <= 3 and bloqueio >=-2:
  print("A bola sobra mas você não conseguiu defeder")
elif decisão == 2 and bloqueio <=-3:
  print("Seus advesarios conseguer pega a bola e fazer uma cesta")
  time2 = +1
else:
  print("invalido")

if time1 >=1 and time1 > time2:
 print("Parabens vocês veceram a partida,rodizio de graça para vocês")
elif time2 >=1 and time2 > time1:
  print("Vocês penderam a partida,voçes pagar o rodizio hoje")
elif time1 == time2:
  print("jogo terminou em empate,todos pagam o rodizio")
else:
  print("invalido")
```
